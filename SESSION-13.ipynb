```python
# 3.Create a Pandas DataFrame for your top 5 favorite Zomato restaurants with columns: 'name', 'rating', 'votes', and 'avg_cost'. Perform bivariate analysis between 'rating' and 'votes' by calculating their correlation coefficient.

import pandas as pd

df = pd.DataFrame({
    'name': ['Pizza Hut', 'Dominos', 'McDonalds', 'Burger King', 'Subway'],
    'rating': [4.2, 4.5, 4.0, 3.8, 4.1],
    'votes': [1200, 1500, 900, 700, 1000],
    'avg_cost': [500, 600, 400, 450, 350]
})

print(df)

correlation = df['rating'].corr(df['votes'])

print("\nCorrelation between Rating and Votes:", round(correlation, 3))
```
